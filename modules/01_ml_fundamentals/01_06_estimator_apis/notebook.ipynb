> **Learner question:** What does a scikit-learn estimator remember before and after `fit`, and how does the same object turn new rows into predictions?

This is a deterministic, synthetic, dimensionless API demonstration. It makes no performance or generalisation claim and does not derive the estimator's internal optimizer.

## S1 — Estimator object and feature matrix

A feature matrix uses rows for samples and columns for features. One feature still means a two-dimensional matrix: `X.shape` is `(4, 1)`, while the target vector `y.shape` is `(4,)`.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

RTOL = 0.0
ATOL = 1e-12
X = np.array([[1.0], [2.0], [3.0], [4.0]], dtype=np.float64)
y = np.array([3.0, 5.0, 7.0, 9.0], dtype=np.float64)
X_new = np.array([[5.0], [6.0]], dtype=np.float64)
print("X shape:", X.shape, "y shape:", y.shape, "X_new shape:", X_new.shape)
```

## S2 — Constructor parameters

Construct the estimator before supplying data. `fit_intercept=True` is a constructor parameter, not a value learned from the rows. `get_params()` exposes this configuration.

```python
estimator = LinearRegression(fit_intercept=True)
print("fit_intercept:", estimator.get_params()["fit_intercept"])
```

## S3 — Learned attributes

Attributes learned from data conventionally end in an underscore: `coef_`, `intercept_`, and `n_features_in_`. They describe post-fit state and are not constructor parameters.

## S4 — The `fit` lifecycle

The lifecycle is `construct → fit(X, y) → inspect learned state → predict(X_new)`. `fit` returns the same estimator object after updating its learned state. The least-squares procedure inside the library is outside this lesson.

```python
fitted = estimator.fit(X, y)
print("fit returned same object:", fitted is estimator)
print("coef_:", fitted.coef_)
print("intercept_:", fitted.intercept_)
print("n_features_in_:", fitted.n_features_in_)
```

## S5 — Hand-compute the fixed line

The frozen table follows `y = 2*x + 1`, so the learned coefficient is `2.0` and the intercept is `1.0`. For the new inputs, `2*5+1=11` and `2*6+1=13`.

## S6 — Predict new rows

`X_new` has shape `(2, 1)`, matching the fitted feature count `n_features_in_=1`. The prediction output can be checked against the hand calculation.

```python
predictions = fitted.predict(X_new)
print("predict(X_new):", predictions)
```

## S7 — Deterministic implementation

The complete example uses explicit `np.float64` arrays, one configured estimator, one `fit` call, and one `predict` call. No random data, preprocessing, evaluation procedure, or model selection is introduced.

## S8 — Verification

The verification contract uses `np.testing.assert_allclose` with `RTOL=0.0` and `ATOL=1e-12` and checks API state as well as predictions.

```python
assert fitted is estimator
assert X.shape == (4, 1) and y.shape == (4,) and X_new.shape == (2, 1)
assert fitted.n_features_in_ == 1
np.testing.assert_allclose(fitted.coef_, [2.0], rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(fitted.intercept_, 1.0, rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(predictions, [11.0, 13.0], rtol=RTOL, atol=ATOL)
print("verified: fit identity, learned state, shapes, and predictions")
```

## S10 — Failure modes

- A one-dimensional `X=[1,2,3,4]` hides the rows-by-features schema.
- Mismatched row and target counts cannot provide one target per sample.
- `X_new` with two columns does not match `n_features_in_=1`.
- Predicting before `fit` requests learned state that does not exist.
- `fit_intercept` is configuration; `intercept_` is learned state.
- Exact toy predictions are not a performance or generalisation claim.
