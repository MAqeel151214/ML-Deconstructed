This lesson uses synthetic, dimensionless teaching values. It introduces supervised learning vocabulary and separates fixed-rule inference, evaluation, and training without making real-world accuracy, generalization, causal, or fairness claims.

## 1. The Problem

What does it mean for a machine-learning model to learn, and how is that different from applying a fixed rule or checking its predictions?

Mitchell's compact vocabulary describes learning using a task **T**, experience **E**, and performance measure **P**: performance on the task improves with experience. This lesson first makes the stages operational: inference applies fixed parameters, evaluation compares predictions with reference targets, and training selects or updates parameters using data and an objective.

## 2. Toy Example or Dataset

A supervised example pairs a feature with a reference target. Here `N=4`, `D=1`, and the design matrix is a two-dimensional column array. All values are synthetic and dimensionless; the exact relationship `y=2x+1` is an artificial construction.

| math index | NumPy row | feature `x_i` | target `y_i` |
|---:|---:|---:|---:|
| 1 | 0 | 1.0 | 3.0 |
| 2 | 1 | 2.0 | 5.0 |
| 3 | 2 | 3.0 | 7.0 |
| 4 | 3 | 4.0 | 9.0 |

We compare two **fixed, manually supplied** rules: Rule 1 `w=2.0,b=1.0` and Rule 2 `w=1.5,b=1.0`. Supplying these values is not training.

```python
import numpy as np

X = np.array([[1.0], [2.0], [3.0], [4.0]], dtype=np.float64)
y = np.array([3.0, 5.0, 7.0, 9.0], dtype=np.float64)
w1 = np.array([2.0], dtype=np.float64)
w2 = np.array([1.5], dtype=np.float64)
b1 = 1.0
b2 = 1.0
X.shape, y.shape, X.dtype, y.dtype
```

## 3. Intuition Before Math

The terms used here are deliberately compact:

| concept | meaning here |
|---|---|
| feature | input variable supplied to a model |
| target/label | reference output paired with an example |
| example | one feature-target pair |
| dataset | finite collection of examples |
| model | parameterized rule family |
| parameter | internal value such as `w` or `b` |
| hyperparameter | external configuration, such as choosing a linear rather than quadratic family |
| prediction | output produced by evaluating a rule |
| inference | applying fixed parameters to inputs |
| training | algorithmically selecting or updating parameters from data and an objective |
| residual/loss | signed discrepancy / non-negative discrepancy summary |
| evaluation | comparing predictions and targets with a stated criterion |

A fixed rule is not automatically a learned model. A model becomes learned only when a training procedure selects or updates its parameters from experience.

## 4. Mathematical Derivation

For one feature, the affine rule is

`y_hat_i = f_theta(x_i) = w x_i + b`.

Across the batch, `y_hat = X @ w + b`, where `(4,1) @ (1,) -> (4,)` and scalar `b` broadcasts. Evaluation uses the frozen residual convention

`e_i = y_i - y_hat_i`, `MAE = (1/N) sum |e_i|`, and `MSE = (1/N) sum e_i^2`.

**Inference:** inputs `X` and fixed `theta` produce `y_hat`; targets are not needed. **Evaluation:** `y_hat`, targets `y`, and metric `P` produce residuals and summaries; parameters remain fixed. **Training:** a data-and-objective procedure selects or updates `theta`; that stage is deferred to 01.05.

## 5. Manual Walkthrough

**Rule 1 (`w=2.0,b=1.0`)**

- `i=1`: `y_hat_1=(2.0)(1.0)+1.0=2.0+1.0=3.0`; `e_1=3.0-3.0=0.0`; `e_1^2=0.00`
- `i=2`: `y_hat_2=(2.0)(2.0)+1.0=4.0+1.0=5.0`; `e_2=5.0-5.0=0.0`; `e_2^2=0.00`
- `i=3`: `y_hat_3=(2.0)(3.0)+1.0=6.0+1.0=7.0`; `e_3=7.0-7.0=0.0`; `e_3^2=0.00`
- `i=4`: `y_hat_4=(2.0)(4.0)+1.0=8.0+1.0=9.0`; `e_4=9.0-9.0=0.0`; `e_4^2=0.00`

`MAE_1=(0+0+0+0)/4=0.0`; `MSE_1=(0+0+0+0)/4=0.0`.

**Rule 2 (`w=1.5,b=1.0`)**

- `i=1`: `y_hat_1=(1.5)(1.0)+1.0=1.5+1.0=2.5`; `e_1=3.0-2.5=+0.5`; `e_1^2=0.25`
- `i=2`: `y_hat_2=(1.5)(2.0)+1.0=3.0+1.0=4.0`; `e_2=5.0-4.0=+1.0`; `e_2^2=1.00`
- `i=3`: `y_hat_3=(1.5)(3.0)+1.0=4.5+1.0=5.5`; `e_3=7.0-5.5=+1.5`; `e_3^2=2.25`
- `i=4`: `y_hat_4=(1.5)(4.0)+1.0=6.0+1.0=7.0`; `e_4=9.0-7.0=+2.0`; `e_4^2=4.00`

`MAE_2=(0.5+1.0+1.5+2.0)/4=5.0/4=1.25`; `MSE_2=(0.25+1.00+2.25+4.00)/4=7.50/4=1.875`.

Rule 1's zero error demonstrates arithmetic on these four synthetic points, not training, generalization, model quality, or causality.

```python
y_hat_1 = X @ w1 + b1
y_hat_2 = X @ w2 + b2
e_1 = y - y_hat_1
e_2 = y - y_hat_2
summary_1 = {"mae": float(np.mean(np.abs(e_1))), "mse": float(np.mean(e_1**2))}
summary_2 = {"mae": float(np.mean(np.abs(e_2))), "mse": float(np.mean(e_2**2))}
y_hat_1, e_1, summary_1, y_hat_2, e_2, summary_2
```

## 6. Teaching Implementation

These transparent functions perform fixed-rule inference and descriptive evaluation only. No estimator, optimizer, random generator, data split, or external ML library is involved.

```python
def predict_linear(X, w, b):
    return X @ w + float(b)

def compute_residuals(y, y_hat):
    return y - y_hat

def summarize_errors(residuals):
    return {"mae": float(np.mean(np.abs(residuals))), "mse": float(np.mean(residuals**2))}

predictions_1 = predict_linear(X, w1, b1)
predictions_2 = predict_linear(X, w2, b2)
errors_1 = compute_residuals(y, predictions_1)
errors_2 = compute_residuals(y, predictions_2)
summarize_errors(errors_1), summarize_errors(errors_2)
```

## 7. Verification

The implementation contract requires exact shapes and dtypes plus tolerance-based value checks. The sign guard confirms that Rule 2's first and last residuals are positive: the rule under-predicts the targets.

```python
ATOL = 1e-12
assert X.shape == (4, 1) and y.shape == (4,)
assert X.dtype == np.float64 and y.dtype == np.float64
np.testing.assert_allclose(predictions_1, [3, 5, 7, 9], rtol=0.0, atol=ATOL)
np.testing.assert_allclose(errors_1, [0, 0, 0, 0], rtol=0.0, atol=ATOL)
np.testing.assert_allclose(predictions_2, [2.5, 4, 5.5, 7], rtol=0.0, atol=ATOL)
np.testing.assert_allclose(errors_2, [0.5, 1, 1.5, 2], rtol=0.0, atol=ATOL)
assert errors_2[0] == 0.5 and errors_2[-1] == 2.0
print("verification: PASS")
```

## 8. Independent Cross-Check

A pure-Python row-by-row calculation and an augmented homogeneous matrix route independently reproduce both rules. Neither pathway calls the teaching functions or a fitted estimator.

```python
for w, b, expected in ((w1, b1, predictions_1), (w2, b2, predictions_2)):
    scalar = []
    residuals = []
    for idx in range(4):
        prediction = float(w[0]) * float(X[idx, 0]) + float(b)
        scalar.append(prediction)
        residuals.append(float(y[idx]) - prediction)
    np.testing.assert_allclose(scalar, expected, rtol=0.0, atol=ATOL)
    np.testing.assert_allclose(residuals, y - expected, rtol=0.0, atol=ATOL)

X_aug = np.column_stack([X, np.ones(4, dtype=np.float64)])
np.testing.assert_allclose(X_aug @ np.array([2.0, 1.0]), predictions_1, rtol=0.0, atol=ATOL)
np.testing.assert_allclose(X_aug @ np.array([1.5, 1.0]), predictions_2, rtol=0.0, atol=ATOL)
print("independent scalar and augmented-matrix checks: PASS")
```

## 10. Failure Modes

- A model is not the dataset: `(X,y)` contains examples; `f_theta` is a rule family.
- Features and targets are not interchangeable: `X` is supplied as input; `y` is the reference output.
- A prediction is not automatically true: Rule 2 differs from `y`.
- Computing `X @ w + b` is inference, not training; manually typed parameters are not learned parameters.
- Parameters `(w,b)` are internal rule values; hyperparameters configure the family or algorithm.
- A perfect fit on four synthetic points does not prove generalization or model quality.
- A prediction does not establish causality.
- Evaluation requires reference targets and a stated criterion; inference does not.
- A residual is not a parameter: `e` is per-example discrepancy, while `theta` contains model coefficients.
- More features do not automatically improve generalization (deferred to 01.03); evaluation context and splits are deferred to 01.04.
