> **Learner question:** How can one straight line choose its slope and intercept to make the squared prediction errors as small as possible?

This is a deterministic, synthetic, dimensionless demonstration. It makes no performance, inference, or generalisation claim.

## S1 — Visible data and line

A one-feature line is `y_hat_i = b_0 + b_1*x_i`. The intercept `b_0` shifts the line vertically; the slope `b_1` controls its tilt.

```python
import numpy as np

RTOL = 0.0
ATOL = 1e-12
x = np.array([1.0, 2.0, 3.0, 4.0], dtype=np.float64)
y = np.array([2.0, 3.0, 5.0, 4.0], dtype=np.float64)
x_new = np.array([5.0, 6.0], dtype=np.float64)
print('x:', x, 'y:', y, 'x_new:', x_new)
```

## S2 — Predictions and residuals

For a candidate line, the residual is observed minus predicted: `r_i = y_i - y_hat_i`. Positive residuals lie above the line; negative residuals lie below it.

## S3 — Least-squares objective

Ordinary least squares minimizes the residual sum of squares:

$$RSS(b_0,b_1)=\sum_i (y_i-b_0-b_1x_i)^2.$$

Squaring prevents positive and negative residuals from cancelling. Here `MSE = RSS / 4`; it is a rescaled summary, not a different fitted line.

## S4 — Centering

The sample means are `x_bar = 2.5` and `y_bar = 3.5`. Centering subtracts these means and lets the slope be computed from co-movement around the mean point.

## S5 — Centered slope

The one-feature closed form is:

$$b_1=\frac{\sum_i(x_i-x_bar)(y_i-y_bar)}{\sum_i(x_i-x_bar)^2}.$$

For this table, the numerator is `4.0` and denominator is `5.0`, so `b_1 = 0.8`.

## S6 — Intercept

The fitted line passes through the mean point, so `b_0 = y_bar - b_1*x_bar = 3.5 - 0.8*2.5 = 1.5`.

## S7 — Complete hand calculation

The fitted rule is `y_hat = 1.5 + 0.8*x`. Its fitted values are `[2.3, 3.1, 3.9, 4.7]`, residuals are `[-0.3, -0.1, 1.1, -0.7]`, `RSS = 1.80`, and `MSE = 0.45`. At new inputs `[5.0, 6.0]`, predictions are `[5.5, 6.3]`.

```python
x_bar = np.mean(x)
y_bar = np.mean(y)
centered_x = x - x_bar
centered_y = y - y_bar
slope = np.sum(centered_x * centered_y) / np.sum(centered_x ** 2)
intercept = y_bar - slope * x_bar
fitted = intercept + slope * x
residuals = y - fitted
rss = np.sum(residuals ** 2)
mse = rss / x.size
predictions = intercept + slope * x_new
print('x_bar:', x_bar, 'y_bar:', y_bar)
print('slope:', slope, 'intercept:', intercept)
print('fitted:', fitted, 'residuals:', residuals)
print('RSS:', rss, 'MSE:', mse, 'new predictions:', predictions)
```

## S8 — Deterministic NumPy implementation

The calculation uses `np.float64`, explicit one-dimensional arrays, centered sums, and separate named quantities so every step remains inspectable.

```python
np.testing.assert_allclose([x_bar, y_bar, slope, intercept, rss, mse], [2.5, 3.5, 0.8, 1.5, 1.8, 0.45], rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(fitted, [2.3, 3.1, 3.9, 4.7], rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(residuals, [-0.3, -0.1, 1.1, -0.7], rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(predictions, [5.5, 6.3], rtol=RTOL, atol=ATOL)
print('NumPy implementation matches the frozen contract.')
```

## S9 — Verification and estimator comparison

The scalar arithmetic is the derivation. A library estimator is a separate consistency check.

```python
from sklearn.linear_model import LinearRegression

estimator = LinearRegression(fit_intercept=True).fit(X := x.reshape(-1, 1), y)
library_predictions = estimator.predict(x_new.reshape(-1, 1))
np.testing.assert_allclose(estimator.coef_, [slope], rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(estimator.intercept_, intercept, rtol=RTOL, atol=ATOL)
np.testing.assert_allclose(library_predictions, predictions, rtol=RTOL, atol=ATOL)
print('LinearRegression agrees:', estimator.coef_, estimator.intercept_, library_predictions)
```

## S11 — Failure modes and exercises

Reversing the residual sign changes labels even though RSS is unchanged after squaring. Using uncentered sums or the wrong denominator does not implement this formula. A displayed loss is not a performance or generalisation guarantee, and new-input predictions are not evidence that extrapolation is safe.

## S12 — References, takeaways, and continuity

Simple OLS chooses the slope from centered co-movement and the intercept from the mean point. For this table, `b_1 = 0.8`, `b_0 = 1.5`, `RSS = 1.8`, and `MSE = 0.45`. The next lesson may extend one feature to multiple features; matrix normal equations and regularization are outside this lesson.
