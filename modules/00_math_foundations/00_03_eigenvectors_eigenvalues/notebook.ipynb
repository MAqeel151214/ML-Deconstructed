## 1. The Problem

A matrix can change both the length and direction of a vector. Are there any non-zero directions that a transformation does **not** turn away from their own lines?

Those special directions are useful mathematical landmarks of a transformation. This lesson studies them in a small, exact 2-D example. It does not make a claim about model quality, feature importance, or causality.

## 2. The Toy Example

All entries below are **dimensionless coordinate counts**. Mathematical vectors are displayed as $(2,1)$ columns; later NumPy code uses a rank-1 array of shape `(2,)` for one vector.

$$
A = \begin{bmatrix}2 & 1\\1 & 2\end{bmatrix},\qquad
v_+ = \begin{bmatrix}1\\1\end{bmatrix},\qquad
v_- = \begin{bmatrix}1\\-1\end{bmatrix},\qquad
u = \begin{bmatrix}2\\1\end{bmatrix}.
$$

We will also use one short sign contrast:

$$
B = \begin{bmatrix}-2 & 0\\0 & 1\end{bmatrix},\qquad
w = \begin{bmatrix}1\\0\end{bmatrix}.
$$

The first matrix is the main example. The second exists only to make a negative scale factor visible.

## 3. Intuition Before Math

An **eigenvector** is not necessarily an arrow that stays unchanged. It is a non-zero arrow whose transformed result stays on the same **line through the origin**. The arrow may stretch, shrink, or reverse.

The number that tells us how it changes is the **eigenvalue**. A positive value preserves arrow orientation; a negative value reverses it. A generic vector can turn to a different line, so not every input vector is an eigenvector.

## 4. Mathematical Derivation

For a square matrix $A$, a non-zero vector $v$ is an eigenvector with eigenvalue $\lambda$ when

$$A v = \lambda v, \qquad v \ne 0.$$

To find the special directions of the primary matrix, let $v=[x,y]^T$ be non-zero. Expanding $A v = \lambda v$ gives

$$
2x+y=\lambda x,\qquad x+2y=\lambda y.
$$

Subtract the second equation from the first:

$$
(2x+y)-(x+2y)=\lambda x-\lambda y
\quad\Longrightarrow\quad
(1-\lambda)(x-y)=0.
$$

There are two cases.

- If $x=y$, then $x\ne0$ and $3x=\lambda x$, so $\lambda=3$ and $v=t[1,1]^T$ for any $t\ne0$.
- If $x\ne y$, then $1-\lambda=0$, so $\lambda=1$. Substituting gives $x+y=0$, so $v=t[1,-1]^T$ for any $t\ne0$.

For this matrix, the two eigen-directions are represented by $[1,1]^T$ with eigenvalue $3$ and $[1,-1]^T$ with eigenvalue $1$. This is a direct coordinate derivation, not a characteristic-polynomial method.

## 5. Manual Walkthrough

### First eigenpair

$$
A v_+ =
\begin{bmatrix}2 & 1\\1 & 2\end{bmatrix}
\begin{bmatrix}1\\1\end{bmatrix}
=
\begin{bmatrix}2\cdot1+1\cdot1\\1\cdot1+2\cdot1\end{bmatrix}
=
\begin{bmatrix}3\\3\end{bmatrix}
=3v_+.
$$

The positive eigenvalue $3$ preserves orientation and triples the arrow length.

### Second eigenpair

$$
A v_- =
\begin{bmatrix}2 & 1\\1 & 2\end{bmatrix}
\begin{bmatrix}1\\-1\end{bmatrix}
=
\begin{bmatrix}2-1\\1-2\end{bmatrix}
=
\begin{bmatrix}1\\-1\end{bmatrix}
=1v_-.
$$

An eigenvalue of $1$ preserves both orientation and length.

### A non-example

$$
A u=
\begin{bmatrix}2 & 1\\1 & 2\end{bmatrix}
\begin{bmatrix}2\\1\end{bmatrix}
=
\begin{bmatrix}5\\4\end{bmatrix}.
$$

If $[5,4]^T$ were a scalar multiple of $[2,1]^T$, their 2-D parallelism scalar would be zero. Instead, $2\cdot4-1\cdot5=3\ne0$, so $u$ is not an eigenvector.

### Negative signed scale

$$
B w=
\begin{bmatrix}-2 & 0\\0 & 1\end{bmatrix}
\begin{bmatrix}1\\0\end{bmatrix}
=
\begin{bmatrix}-2\\0\end{bmatrix}
=-2w.
$$

The vector stays on the horizontal line, reverses orientation, and doubles in length.

## 6. Teaching Implementation

The code below checks **given** candidates. It does not use a numerical eigensolver to discover eigenpairs. Before multiplying, it checks that the matrix is square and the rank-1 vector has compatible length.

````python
import numpy as np

A = np.array([[2, 1], [1, 2]], dtype=int)
v_plus = np.array([1, 1], dtype=int)
v_minus = np.array([1, -1], dtype=int)
u = np.array([2, 1], dtype=int)

B = np.array([[-2, 0], [0, 1]], dtype=int)
w = np.array([1, 0], dtype=int)

def require_square_matrix(matrix):
    if matrix.ndim != 2 or matrix.shape[0] != matrix.shape[1]:
        raise ValueError(f"matrix must have shape (n, n), got {matrix.shape}")

def require_compatible_vector(matrix, vector):
    if vector.ndim != 1 or vector.shape[0] != matrix.shape[0]:
        raise ValueError(
            f"vector shape {vector.shape} is incompatible with {matrix.shape}"
        )

def is_exact_eigenpair(matrix, vector, eigenvalue):
    require_square_matrix(matrix)
    require_compatible_vector(matrix, vector)
    if not np.any(vector != 0):
        return False
    return bool(np.array_equal(matrix @ vector, eigenvalue * vector))

print("A.shape:", A.shape)
print("v_plus.shape:", v_plus.shape)
print("A @ v_plus:", A @ v_plus)
````

## 7. Verification of Fixed Teaching Calculations

The assertions below verify every fixed result from the manual walkthrough. Exact integer arrays let us use exact equality here.

````python
zero = np.array([0, 0], dtype=int)

assert A.shape == (2, 2)
assert v_plus.shape == v_minus.shape == u.shape == (2,)
assert B.shape == (2, 2)
assert w.shape == zero.shape == (2,)

np.testing.assert_array_equal(A @ v_plus, [3, 3])
np.testing.assert_array_equal(A @ v_minus, [1, -1])
np.testing.assert_array_equal(A @ u, [5, 4])
np.testing.assert_array_equal(B @ w, [-2, 0])

assert is_exact_eigenpair(A, v_plus, 3)
assert is_exact_eigenpair(A, v_minus, 1)
assert is_exact_eigenpair(B, w, -2)
assert not is_exact_eigenpair(A, zero, 0)

parallelism_scalar = u[0] * (A @ u)[1] - u[1] * (A @ u)[0]
assert parallelism_scalar == 3

print("PASS")
````

## 8. Independent Cross-Check

Matrix multiplication with `@` is concise. As an independent check, recompute each 2-D output from its two scalar row-dot-products and compare it with NumPy's result.

````python
def scalar_row_dot_product(matrix, vector):
    if matrix.shape != (2, 2) or vector.shape != (2,):
        raise ValueError("expected a (2, 2) matrix and a (2,) vector")
    return np.array([
        matrix[0, 0] * vector[0] + matrix[0, 1] * vector[1],
        matrix[1, 0] * vector[0] + matrix[1, 1] * vector[1],
    ], dtype=int)

for matrix, vector in ((A, v_plus), (A, v_minus), (A, u), (B, w)):
    np.testing.assert_array_equal(matrix @ vector, scalar_row_dot_product(matrix, vector))

print("PASS: NumPy and scalar row-dot-products agree")
````

## 10. Failure Modes & Pitfalls

- **An eigenvector is unchanged.** Not necessarily: $A v_+=3v_+$ changes length, and $B w=-2w$ changes length and orientation.
- **Every input vector is an eigenvector.** No: $A u=[5,4]^T$ is not parallel to $u=[2,1]^T$.
- **The zero vector is an eigenvector.** No: $A0=\lambda0$ is true for every $\lambda$, so it supplies no direction.
- **A negative eigenvalue disqualifies the vector.** No: the negative sign reverses orientation on the same line.
- **Every real 2-D matrix has two real eigen-directions.** Not guaranteed; a 90-degree rotation is a boundary example, not a case solved here.

### Spot the bug

```python
def is_eigenpair(A, v, eigenvalue):
    return np.array_equal(A @ v, eigenvalue * v)
```

This helper incorrectly accepts `v = np.array([0, 0])` for every eigenvalue. The definition requires a non-zero vector, so add a zero-vector guard before the equality check.

## 11. Exercises

### 1. Conceptual recognition
Classify as true or false: (a) $v_+$ is an eigenvector because $A v_+=3v_+$; (b) $u$ is an eigenvector because $A u$ is 2-D; (c) $v_-$ is not an eigenvector because its eigenvalue is $1$; (d) $w$ is not an eigenvector of $B$ because its output reverses.

<details><summary>Answer</summary>

(a) true. (b) false: the output is not a scalar multiple. (c) false: $1v_-$ satisfies the definition. (d) false: $-2w$ is still a scalar multiple.
</details>

### 2. Exact scalar-multiple test
Why can no scalar $c$ make $[5,4]^T=c[2,1]^T$? What does this prove about $u$?

<details><summary>Answer</summary>

The first coordinate requires $c=5/2$, while the second requires $c=4$. No one scalar works, so $u$ is not an eigenvector of $A$.
</details>

### 3. Same eigen-line, different representative
Let $\tilde v=-2v_+$. Use $A v_+=3v_+$ to find $A\tilde v$ and its eigenvalue.

<details><summary>Answer</summary>

$A\tilde v=A(-2v_+)=-2(A v_+)=3(-2v_+)=3\tilde v$. Its eigenvalue is still $3$.
</details>

### 4. Implementation: shapes
Why is `A @ v_plus` valid for these teaching arrays? State the required shape of each array.

<details><summary>Answer</summary>

`A` has shape `(2, 2)` and `v_plus` has shape `(2,)`. More generally, a square `(n, n)` matrix needs a rank-1 vector of shape `(n,)`.
</details>

### 5. Implementation: zero-vector guard
Write one Boolean condition that rejects the zero vector before testing the eigenpair equation.

<details><summary>Answer</summary>

`if not np.any(vector != 0): return False` rejects a vector whose entries are all zero.
</details>

### 6. Extension: signed scale
A non-zero vector satisfies $C r=(-1/2)r$. Describe the output arrow without inventing a matrix.

<details><summary>Answer</summary>

It stays on the same line, reverses orientation, and has half the input length.
</details>
