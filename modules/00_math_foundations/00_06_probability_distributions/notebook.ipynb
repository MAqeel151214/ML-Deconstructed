## 1. The Problem

A probability distribution gives a structured answer to two questions: which outcomes are possible, and how much probability belongs to each outcome? How can a small table tell us what is likely, what the average outcome is, and how spread out the outcomes are?

We will use one biased coin flip as a complete discrete example, then make a short geometric contrast with a continuous density.

## 2. Toy Example or Dataset

Let `X` be the number of heads from one flip of a biased coin.

```text
sample space: {0, 1}
X=0: tails
X=1: heads
P(X=0) = 0.4
P(X=1) = 0.6
```

`X` is the random variable. Lowercase `x` denotes one particular outcome. The event “heads” is `{X=1}`.

````
import numpy as np

outcomes = np.array([0.0, 1.0], dtype=np.float64)
probabilities = np.array([0.4, 0.6], dtype=np.float64)

print("outcomes:", outcomes)
print("probabilities:", probabilities)
print("shapes:", outcomes.shape, probabilities.shape)
````

## 3. Intuition Before Math

The probability mass function (PMF) assigns probability to each discrete outcome. Here, outcome `1` is more likely than outcome `0`, so a weighted average should be closer to `1`.

The expectation is a weighted average, not necessarily one of the possible outcomes and not automatically the most likely outcome. We will also calculate a separate quantity for spread.

## 4. Mathematical Derivation

For a discrete random variable:

```text
E[X] = Σ x P(X=x)
E[X²] = Σ x² P(X=x)
Var(X) = E[X²] - E[X]²
σ = sqrt(Var(X))
```

A valid PMF has non-negative entries and all entries sum to one. The frozen table satisfies:

```text
0.4 >= 0
0.6 >= 0
0.4 + 0.6 = 1.0
```

Variance measures squared spread around the mean; standard deviation returns to the original outcome units.

## 5. Manual Walkthrough

### PMF and event probability

```text
x=0: P(X=0)=0.4
x=1: P(X=1)=0.6

P(X=1) = 0.6
```

### Expectation

```text
E[X] = (0)(0.4) + (1)(0.6)
     = 0.0 + 0.6
     = 0.6
```

### Second moment

```text
E[X²] = (0²)(0.4) + (1²)(0.6)
      = 0.0 + 0.6
      = 0.6
```

Because `x²=x` for both outcomes `0` and `1`, `E[X²]=E[X]` is a useful sanity check.

### Variance and standard deviation

```text
Var(X) = 0.6 - (0.6)²
       = 0.6 - 0.36
       = 0.24

σ = sqrt(0.24)
  = 0.4898979485566356...
```

The mean is `0.6`, the variance is `0.24`, and the standard deviation is approximately `0.49`.

````
expectation = np.sum(outcomes * probabilities)
second_moment = np.sum(outcomes**2 * probabilities)
variance = second_moment - expectation**2
std_deviation = np.sqrt(variance)

print("E[X] =", expectation)
print("E[X^2] =", second_moment)
print("Var(X) =", variance)
print("sigma =", std_deviation)
````

## 6. Teaching Implementation

The two arrays align term by term: each outcome is multiplied by its probability. The implementation uses `float64` because the frozen probabilities and derived values are decimal quantities.

```python
expectation = np.sum(outcomes * probabilities)
second_moment = np.sum(outcomes**2 * probabilities)
variance = second_moment - expectation**2
std_deviation = np.sqrt(variance)
```

No random flip is needed: this is a theoretical distribution specified directly by its PMF.

````
assert outcomes.dtype == np.float64
assert probabilities.dtype == np.float64
assert outcomes.shape == (2,)
assert probabilities.shape == (2,)
assert np.all(probabilities >= 0)
np.testing.assert_allclose(probabilities.sum(), 1.0, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(expectation, 0.6, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(second_moment, 0.6, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(variance, 0.24, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(
    std_deviation, 0.4898979485566356, rtol=1e-12, atol=1e-12
)
print("PASS")
````

## 7. Verification

The PMF axioms and every frozen numerical result are checked with assertions. These checks validate deterministic teaching arithmetic; they are not statistical tests.

````
expected_outcomes = np.array([0.0, 1.0], dtype=np.float64)
expected_probabilities = np.array([0.4, 0.6], dtype=np.float64)
np.testing.assert_allclose(outcomes, expected_outcomes, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(probabilities, expected_probabilities, rtol=1e-12, atol=1e-12)
assert np.all(probabilities >= 0)
np.testing.assert_allclose(probabilities.sum(), 1.0, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(probabilities[1], 0.6, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(expectation, 0.6, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(second_moment, 0.6, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(variance, 0.24, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(std_deviation, 0.4898979485566356, rtol=1e-12, atol=1e-12)
print("verification PASS")
````

## 8. Independent Cross-Check

Recompute variance through a different formula, using weighted squared deviations:

```text
Σ((x - E[X])² P(X=x))
```

For the frozen table:

```text
(0-0.6)²(0.4) = 0.144
(1-0.6)²(0.6) = 0.096
0.144 + 0.096 = 0.24
```

This independently agrees with `E[X²]-E[X]²`.

````
direct_squared_deviations = (outcomes - expectation) ** 2
direct_variance = np.sum(direct_squared_deviations * probabilities)
np.testing.assert_allclose(direct_variance, 0.24, rtol=1e-12, atol=1e-12)
print("independent variance cross-check:", direct_variance)
````

## 10. Failure Modes

- **Negative probability:** every PMF entry must be non-negative.
- **Total not equal to one:** a complete PMF must sum to one.
- **Expectation as an outcome:** `E[X]=0.6` is a weighted average, not a required outcome.
- **Expectation as the most likely outcome:** `1` is the most likely outcome here; `0.6` is the average.
- **Variance as the mean:** the mean is `0.6`; the variance is `0.24`.
- **PDF height as probability:** for `Uniform(0,1)`, height `1` is density. The interval `[0.25,0.75]` has probability `0.5` because its area is `0.5`.
- **Exact continuous point:** in this continuous example, `P(X=c)=0`.
- **Synthetic data as evidence:** this toy PMF does not describe real behaviour, causality, or model quality.

### Spot the bug

```text
E[X] = 0.6
therefore P(X=0.6) = 1
```

This is wrong: the possible outcomes are `0` and `1`; `0.6` is their weighted average.
